```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta

fake = Faker('en_IN')
np.random.seed(42)
random.seed(42)

print("Libraries loaded successfully!")
```

```python
# Product categories and their products
product_data = {
    'Dairy': [
        ('Amul Milk', 'Amul', 26, 'Litre'),
        ('Mother Dairy Curd', 'Mother Dairy', 45, 'Kg'),
        ('Amul Butter', 'Amul', 55, 'Piece'),
        ('Amul Paneer', 'Amul', 89, 'Piece'),
        ('Nestle Yogurt', 'Nestle', 35, 'Piece')
    ],
    'Snacks': [
        ('Lays Classic', 'Lays', 20, 'Piece'),
        ('Kurkure Masala', 'Kurkure', 20, 'Piece'),
        ('Britannia Biscuit', 'Britannia', 30, 'Piece'),
        ('Haldirams Namkeen', 'Haldirams', 60, 'Piece'),
        ('Too Yumm Chips', 'Too Yumm', 25, 'Piece')
    ],
    'Beverages': [
        ('Coca Cola 500ml', 'Coca Cola', 40, 'Piece'),
        ('Tropicana Juice', 'Tropicana', 99, 'Piece'),
        ('Red Bull', 'Red Bull', 125, 'Piece'),
        ('Bisleri Water', 'Bisleri', 20, 'Piece'),
        ('Nescafe Coffee', 'Nescafe', 299, 'Piece')
    ],
    'Personal Care': [
        ('Dove Shampoo', 'Dove', 179, 'Piece'),
        ('Colgate Toothpaste', 'Colgate', 89, 'Piece'),
        ('Dettol Soap', 'Dettol', 45, 'Piece'),
        ('Nivea Cream', 'Nivea', 199, 'Piece'),
        ('Pantene Conditioner', 'Pantene', 199, 'Piece')
    ],
    'Staples': [
        ('Aashirvaad Atta', 'Aashirvaad', 280, 'Kg'),
        ('India Gate Rice', 'India Gate', 320, 'Kg'),
        ('Tata Salt', 'Tata', 25, 'Kg'),
        ('Fortune Oil', 'Fortune', 180, 'Litre'),
        ('Tata Dal', 'Tata', 150, 'Kg')
    ],
    'Frozen Food': [
        ('McCain Fries', 'McCain', 180, 'Piece'),
        ('Godrej Chicken', 'Godrej', 350, 'Piece'),
        ('Amul Ice Cream', 'Amul', 130, 'Piece'),
        ('Haldirams Paratha', 'Haldirams', 89, 'Piece'),
        ('Venky Nuggets', 'Venky', 220, 'Piece')
    ]
}

# Create products dataframe
all_products = []

for category, items in product_data.items():
    for product_name, brand, price, unit in items:
        all_products.append([len(all_products)+1, product_name, category, brand, price, unit])

df_products = pd.DataFrame(all_products, columns=['product_id', 'product_name', 'category', 'brand', 'selling_price', 'unit'])
df_products['product_id'] = 'P' + df_products['product_id'].astype(str).str.zfill(3)

print(f"Products table created: {df_products.shape}")
print(df_products.head(10))
```

```python
df_products.to_csv('data/products.csv', index=False)
print("products.csv saved successfully!")
print(f"Total products: {len(df_products)}")
```

```python
# Warehouse data
warehouses_data = {
    'warehouse_id': ['W001', 'W002', 'W003', 'W004', 'W005', 'W006', 'W007', 'W008', 'W009', 'W010'],
    'city': ['Bangalore', 'Delhi', 'Mumbai', 'Hyderabad', 'Pune', 'Chennai', 'Kolkata', 'Ahmedabad', 'Bangalore', 'Delhi'],
    'area': ['Koramangala', 'Connaught Place', 'Andheri', 'Banjara Hills', 'Hinjewadi', 'Anna Nagar', 'Salt Lake', 'Navrangpura', 'Whitefield', 'Rohini'],
    'capacity': [5000, 6000, 5500, 4500, 4000, 4800, 4200, 3800, 5200, 5800],
    'utilization_rate': [0.82, 0.91, 0.78, 0.85, 0.72, 0.88, 0.76, 0.69, 0.83, 0.90]
}

df_warehouses = pd.DataFrame(warehouses_data)
print(f"Warehouses table created: {df_warehouses.shape}")
print(df_warehouses)
```

```python
df_warehouses.to_csv('data/warehouses.csv', index=False)
print("warehouses.csv saved successfully!")
print(f"Total warehouses: {len(df_warehouses)}")
```

```python
# Inventory data generation
inventory_rows = []
inventory_id = 1

for _, warehouse in df_warehouses.iterrows():
    for _, product in df_products.iterrows():
        
        # Stock level based on category
        if product['category'] in ['Dairy', 'Beverages']:
            stock = np.random.randint(50, 500)    # High demand - lower stock
        elif product['category'] in ['Staples']:
            stock = np.random.randint(100, 800)   # Stable demand - higher stock
        else:
            stock = np.random.randint(30, 400)    # Normal stock
        
        # Reorder level - 20% of max stock
        reorder_level = int(stock * 0.2)
        
        # Last restock date
        last_restock = fake.date_between(start_date='-30d', end_date='today')
        
        inventory_rows.append({
            'inventory_id': f'I{inventory_id:04d}',
            'product_id': product['product_id'],
            'warehouse_id': warehouse['warehouse_id'],
            'stock_available': stock,
            'reorder_level': reorder_level,
            'last_restock_date': last_restock
        })
        inventory_id += 1

df_inventory = pd.DataFrame(inventory_rows)
print(f"Inventory table created: {df_inventory.shape}")
print(df_inventory.head(10))
```

```python
df_inventory.to_csv('data/inventory.csv', index=False)
print("inventory.csv saved successfully!")
print(f"Total inventory records: {len(df_inventory)}")
```

```python
# Order generation settings
n_orders = 100000
start_date = datetime(2023, 1, 1)
end_date = datetime(2023, 12, 31)

order_rows = []

for i in range(n_orders):
    # Random date generation
    random_days = np.random.randint(0, 365)
    order_date = start_date + timedelta(days=random_days)
    
    # Weekend demand spike - more orders on weekends
    if order_date.weekday() >= 5:  # Saturday = 5, Sunday = 6
        if np.random.random() < 0.3:  # 30% extra orders on weekends
            continue
    
    # Random product and warehouse
    product = df_products.sample(1).iloc[0]
    warehouse = df_warehouses.sample(1).iloc[0]
    
    # Quantity based on category
    if product['category'] in ['Dairy', 'Beverages']:
        quantity = np.random.randint(1, 5)
    else:
        quantity = np.random.randint(1, 3)
    
    # Order amount
    order_amount = quantity * product['selling_price']
    
    # Order status - some orders cancelled due to stockout
    status_prob = np.random.random()
    if status_prob < 0.08:        # 8% stockout/cancelled
        order_status = 'Cancelled'
    elif status_prob < 0.12:      # 4% pending
        order_status = 'Pending'
    else:                          # 88% delivered
        order_status = 'Delivered'
    
    order_rows.append({
        'order_id': f'O{i+1:06d}',
        'product_id': product['product_id'],
        'warehouse_id': warehouse['warehouse_id'],
        'quantity': quantity,
        'order_date': order_date.strftime('%Y-%m-%d'),
        'order_amount': order_amount,
        'order_status': order_status
    })

df_orders = pd.DataFrame(order_rows)
print(f"Orders table created: {df_orders.shape}")
print(df_orders.head(10))
```

```python
df_orders.to_csv('data/orders.csv', index=False)
print("orders.csv saved successfully!")
print(f"Total orders: {len(df_orders)}")
print(f"\nOrder Status Distribution:")
print(df_orders['order_status'].value_counts())
```

```python
# Only delivered and pending orders get delivery records
df_delivered = df_orders[df_orders['order_status'].isin(['Delivered', 'Pending'])]

delivery_rows = []

for i, (_, order) in enumerate(df_delivered.iterrows()):
    order_date = datetime.strptime(order['order_date'], '%Y-%m-%d')
    
    # Expected delivery = 1-2 days after order
    expected_days = np.random.randint(1, 3)
    expected_date = order_date + timedelta(days=expected_days)
    
    # Actual delivery - some delayed
    delay_prob = np.random.random()
    if delay_prob < 0.15:  # 15% delayed deliveries
        delay_days = np.random.randint(1, 5)
        delivery_status = 'Delayed'
    else:
        delay_days = 0
        delivery_status = 'On-Time'
    
    actual_delivery_date = expected_date + timedelta(days=delay_days)
    
    delivery_rows.append({
        'delivery_id': f'D{i+1:06d}',
        'order_id': order['order_id'],
        'expected_date': expected_date.strftime('%Y-%m-%d'),
        'delivery_date': actual_delivery_date.strftime('%Y-%m-%d'),
        'delivery_status': delivery_status,
        'delay_days': delay_days
    })

df_deliveries = pd.DataFrame(delivery_rows)
print(f"Deliveries table created: {df_deliveries.shape}")
print(df_deliveries.head(10))
```

```python
df_deliveries.to_csv('data/deliveries.csv', index=False)
print("deliveries.csv saved successfully!")
print(f"Total delivery records: {len(df_deliveries)}")
print(f"\nDelivery Status Distribution:")
print(df_deliveries['delivery_status'].value_counts())
```
